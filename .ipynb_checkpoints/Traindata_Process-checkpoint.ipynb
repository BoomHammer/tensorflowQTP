```python
import pandas as pd
import numpy as np
import csv
```

```python
data_train=pd.read_csv('F:/TensorFlow/traindata20230508原数据.csv',header=0,encoding='utf_8')
data_train.info()
data_train.ClassName.unique()
```

```python
pd.set_option('display.max_rows', None)
class_counts = data_train['ClassName'].value_counts()
print(class_counts)
```

```python
num_classes = data_train['ClassName'].nunique()
print(num_classes)
```

```python
# 统计每个类别的样本数量
class_counts = data_train['ClassName'].value_counts()

# 将类别数量映射回原表，并作为新的一列插入
data_train['ClassCount'] = data_train['ClassName'].map(class_counts)

# 保存结果到新的CSV文件
data_train.to_csv("F:/TensorFlow/traindata20230508原数据+count.csv", index=False,encoding='utf_8')
```
