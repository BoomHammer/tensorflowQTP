```python
import rasterio
import numpy as np
import pandas as pd
from collections import defaultdict
import math

def calculate_area_and_append(tif_path, csv_path, output_csv_path):
    print("开始读取TIFF影像并统计面积...")
    area_sums = defaultdict(float)
    
    with rasterio.open(tif_path) as src:
        is_geo = src.crs.is_geographic
        print(f"自动检测到坐标系: {'地理坐标系 (例如WGS1984, 像素单位为度)' if is_geo else '投影坐标系 (像素单位为米)'}")
        print(f"当前影像 CRS: {src.crs}")

        if is_geo:
            # === 已替换为严密的 WGS1984 椭球体面积计算公式 ===
            a = 6378137.0 # WGS84 长半轴 (米)
            b = 6356752.314245 # WGS84 短半轴 (米)
            e = math.sqrt(1 - (b**2 / a**2)) # 偏心率
            dx_deg = src.res[0] # X方向分辨率(经度跨度，度)
            
            def get_ellipsoid_area_to_equator(lat_deg):
                # 计算赤道到指定纬度、经度跨度为dx_deg的椭球面面积(平方米)
                lat_rad = math.radians(lat_deg)
                sin_lat = math.sin(lat_rad)
                term1 = sin_lat / (1 - (e**2) * (sin_lat**2))
                term2 = (1 / (2 * e)) * math.log((1 + e * sin_lat) / (1 - e * sin_lat))
                return (dx_deg / 360.0) * 2 * math.pi * (a**2) * (1 - e**2) * 0.5 * (term1 + term2)
                
            row_areas_sqkm = np.zeros(src.height)
            transform = src.transform
            
            for row in range(src.height):
                top_lat = transform[5] + row * transform[4]
                bottom_lat = top_lat + transform[4]
                # 计算两根纬线之间的面积差并转换为平方千米
                area_sqm = abs(get_ellipsoid_area_to_equator(top_lat) - get_ellipsoid_area_to_equator(bottom_lat))
                row_areas_sqkm[row] = area_sqm / 1_000_000.0
            # ===================================================
        else:
            res_x, res_y = src.res
            pixel_area_sqkm = (res_x * res_y) / 1_000_000.0

        for block_index, window in src.block_windows(1):
            block_data = src.read(1, window=window)
            
            if is_geo:
                window_row_areas = row_areas_sqkm[window.row_off : window.row_off + window.height]
                area_grid = window_row_areas[:, np.newaxis]
                area_grid_full = np.broadcast_to(area_grid, block_data.shape)
                
                unique_values = np.unique(block_data)
                for val in unique_values:
                    area_sums[val] += np.sum(area_grid_full[block_data == val])
            else:
                unique_values, counts = np.unique(block_data, return_counts=True)
                for val, count in zip(unique_values, counts):
                    area_sums[val] += count * pixel_area_sqkm

    print("面积统计完成！开始更新CSV字典表...")
    try:
        df = pd.read_csv(csv_path, encoding='gbk')
    except UnicodeDecodeError:
        df = pd.read_csv(csv_path, encoding='utf-8')

    df['面积(平方千米)'] = df['Value'].map(area_sums).fillna(0)
    df.to_csv(output_csv_path, index=False, encoding='utf-8-sig')
    print(f"处理成功！结果已保存至: {output_csv_path}")

if __name__ == "__main__":
    INPUT_TIF = r"F:\新疆文章\面积统计\tr20250701_clcd_XJonly.tif"
    INPUT_CSV = r"F:\新疆文章\面积统计\20250701字典.csv"
    OUTPUT_CSV = r"F:\新疆文章\面积统计\20250701面积.csv"
    calculate_area_and_append(INPUT_TIF, INPUT_CSV, OUTPUT_CSV)
```
