```python
import tensorflow as tf
```

### 载入数据
MNIST是NIST数据库的一个子集。MNIST数据库的官方下载网址为：  
http://yann.lecun.com/exdb/mnist/  
tensorflow2.0 keras中已经集成了mnist 可以直接调用

```python
mnist = tf.keras.datasets.mnist
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

### 数据的形状
训练数据60000条，测试数据10000条，每张28×28

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

### 看一下数据及标签

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(x_train[i])
    plt.xlabel(y_train[i])
plt.show()
```

### 数据预处理
首先对label进行one-hot处理，之后用tf2.0自带的dataAPI进行打包，组合成train 与label的对应数据集。  
我们下载下来的数据集的张量形式为三维的，[x,28,28]但在CNN中数据张量形式为四维的，  
[batch,hight,width,channels],所以需要对原始数据进行增维处理。

```python
import numpy as np
x_train=tf.expand_dims(x_train,-1) #-1表示增加的是第四维
x_test=tf.expand_dims(x_test,-1)
y_train=np.float32(tf.keras.utils.to_categorical(y_train,num_classes=10))
y_test=np.float32(tf.keras.utils.to_categorical(y_test,num_classes=10))
```

### 模型构建

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.BatchNormalization(input_shape=[28, 28,1]),
    tf.keras.layers.Conv2D(32,3,padding='SAME',activation=tf.nn.relu),
    tf.keras.layers.Conv2D(64,3,padding='SAME',activation=tf.nn.relu),
    tf.keras.layers.MaxPool2D(strides=[1,1]),
    tf.keras.layers.Conv2D(128,3,padding='SAME',activation=tf.nn.relu),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(512,activation=tf.nn.relu),
    tf.keras.layers.Dense(128,activation=tf.nn.relu),
    tf.keras.layers.Dense(10,activation=tf.nn.softmax)
])
```

但第一层input_shape必须与输入（x）大小一致，本实验中为[28, 28,1]，  
最后一层节点数必须与输出（类别数量）一致，本实验中为10

##### 本模型中使用到的层有：
BatchNormalization、Conv2D二维卷积层、MaxPool2D二维最大值池化层、Flatten压平、Dense全连接层

### 编译模型
##### optimizer优化器，常用优化器有：
name='GradientDescent'（梯度下降算法）；  
name='Momentum'（动量梯度下降算法）；  
name='Adam'（自适应矩估计算法）；
##### loss损失函数，常用损失函数有：
loss='binary_crossentropy'，（交叉熵损失函数（二分类））；  
loss='categorical_crossentropy'（分类交叉熵函数（多分类one-hot编码））；  
loss='sparse_categorical_crossentropy'（分类交叉熵函数（多分类顺序编码））；

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
print(model.summary())
```

### 训练模型
epochs训练轮数  
batch_size一次训练所选取的样本数，一定范围内越大越好  
大batch_size需要更高的算力，batch_size过小，梯度在最小值附近来回震荡，模型难以收敛
实验表现最好的batch size一般处于2~32之间  

与batch_size相对应的是学习率learning rate，lr越大学习越快，越小越慢。lr过大会导致梯度在最小值附近来回震荡，模型难以收敛

```python
model.fit(x_train,y_train, epochs=10, batch_size=512)
```

callbacks回调函数，记录训练模型的内在状态和统计。  
在训练实时调整学习率、使用Tensorboard实现可视化等情况下会用到。

```python
print(model.evaluate(x_test, y_test))
```
