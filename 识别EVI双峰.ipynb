```python
import pandas as pd
import numpy as np
from scipy import signal
```

```python
all_data=pd.read_csv('F:/TensorFlow/extract2piont_20240401_filtered.csv',header=0,encoding='utf_8')
all_data.head()
```

```python
for idx,row in all_data.iterrows():
    ser = row.iloc[16:]
    peakind = signal.find_peaks_cwt(ser, np.arange(1,6))
    row['peak_num'] = len(ser[peakind]) - 1
    row['peak_names'] = ser[peakind].index.values[1:]
    pd.DataFrame([row]).to_csv('F:/TensorFlow/yellowriver_cornwheat_peak.csv',mode='a', sep=',', header=(idx < 1), index=None, encoding='utf-8')
```
