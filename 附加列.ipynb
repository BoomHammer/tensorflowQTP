```python
import pandas as pd

def merge_csv_files():
    # 1. 定义文件名 (前面加 r 防止路径中的 \t 被转义)
    file_verify = r"G:\Eurasia\蒙古验证.csv"
    file_mapping = r"G:\Eurasia\traindata20260320-mapping.csv"
    output_file = r"G:\Eurasia\蒙古验证_已附加合并列.csv"

    try:
        # 2. 读取 CSV 文件
        print(f"正在读取 {file_verify} 和 {file_mapping}...")
        df_verify = pd.read_csv(file_verify)
        df_mapping = pd.read_csv(file_mapping)

        # 3. 提取 mapping 表格中我们需要的列 (在这里加上了 'Alliance')
        df_mapping_subset = df_mapping[['num', 'Formation', 'Alliance']]

        # 4. 执行合并 (Left Join)
        print("正在进行数据匹配和合并...")
        result_df = pd.merge(df_verify, df_mapping_subset, left_on='prediction', right_on='num', how='left')

        # 5. 清理多余的列（删掉作为桥梁的 num 列）
        if 'num' in result_df.columns:
            result_df = result_df.drop(columns=['num'])

        # 6. 保存为新的 CSV 文件
        result_df.to_csv(output_file, index=False, encoding='utf-8-sig')
        print(f"处理完成！结果已成功保存为: {output_file}")

    except FileNotFoundError as e:
        print(f"错误: 找不到文件。请确保路径正确。\n详细信息: {e}")
    except KeyError as e:
        print(f"错误: 找不到指定的列名。请检查 CSV 文件中是否确实存在 {e} 列。注意大小写。")
    except Exception as e:
        print(f"发生未知错误: {e}")

if __name__ == "__main__":
    merge_csv_files()
```
