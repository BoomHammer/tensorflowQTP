### 为表格增加一列各类别的数量统计

```python
import pandas as pd

# 加载数据
data_train = pd.read_csv("F:/TensorFlow/traindata20241213.csv", header=0, encoding="utf-8")

# 确保类别字段是字符串，并清理多余字符
data_train['ClassName'] = data_train['ClassName'].astype(str).str.strip()
data_train['ClassName'] = data_train['ClassName'].str.replace(r'\s+', '', regex=True)  # 移除所有空白字符
data_train['ClassName'] = data_train['ClassName'].str.normalize('NFKC')  # 标准化中文字符

# 统计每个类别的样本数量
class_counts = data_train['ClassName'].value_counts()

# 将类别数量映射回原表，并作为新的一列插入
data_train['ClassCount'] = data_train['ClassName'].map(class_counts)

# 检查空值行并输出日志
unmatched = data_train[data_train['ClassCount'].isnull()]
if not unmatched.empty:
    print("以下类别未被匹配：")
    print(unmatched['ClassName'].unique())

# 保存结果到新的CSV文件
data_train.to_csv("F:/TensorFlow/traindata20241213+count.csv", encoding="utf-8", index=False)
```

### 为表格增加num列

```python
import re

# 加载数据
data_train = pd.read_csv("F:/TensorFlow/traindata20241213删除数量过少样本.csv", header=0, encoding="utf-8-sig")
# 加载类别编码文件
class_mapping = pd.read_csv("F:/TensorFlow/traindata20241213字典.csv", header=0, encoding="utf-8")

# 调整后的 PureChinese 提取函数
def extract_all_chinese(text):
    if isinstance(text, str):
        # 使用正则表达式提取所有的中文字符，包括括号内的中文
        matches = re.findall(r'[\u4e00-\u9fa5]+', text)
        return ''.join(matches)
    return None

# 更新主表和映射表的 PureChinese
data_train['PureChinese'] = data_train['ClassName'].apply(extract_all_chinese)
class_mapping['PureChinese'] = class_mapping['ClassName'].apply(extract_all_chinese)
```

```python

# 构建映射
class_code_map = dict(zip(class_mapping['PureChinese'], class_mapping['num']))
data_train['num'] = data_train['PureChinese'].map(class_code_map)

# 未匹配处理
unmatched = data_train[data_train['num'].isnull()]
if not unmatched.empty:
    print("以下类别未匹配到编码，请检查：")
    print(unmatched[['ClassName', 'PureChinese']].drop_duplicates())

# 保存结果
data_train.to_csv("F:/TensorFlow/traindata20241213.csv", encoding="utf-8-sig", index=False)
```
