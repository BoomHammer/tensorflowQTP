```python
import pandas as pd
```

```python
df = pd.read_csv('F:/TensorFlow/xinjiang/traindata20250626_3.csv',header=0,encoding='=UTF-8')
alliance_counts = df['Alliance'].value_counts()
print(alliance_counts.to_string())
```

```python
# 获取样本数>=5的类别列表
valid_categories = alliance_counts[alliance_counts >= 5].index
# 根据有效类别过滤数据
new_df = df[df['Alliance'].isin(valid_categories)].copy()
```

```python
uni=new_df.Alliance.unique()
print('共计'+ str(len(uni)) +'个类别')
i = 0
mapping = {}
for item in uni:
    mapping[item] = i
    i+=1
print(mapping)

new_df['num'] = new_df['Alliance'].map(mapping)
new_df.head
```

```python
new_df.to_csv(path_or_buf='G:/Eurasia/traindata20260320.csv', encoding='utf_8_sig')
```

```python
required_columns = ['Alliance', 'num', 'Formation']
summary_df = new_df[required_columns].drop_duplicates()
result_df = summary_df.sort_values(by='num').reset_index(drop=True)
print(result_df)
```

```python
result_df.to_csv(path_or_buf='G:/Eurasia/traindata20260320-mapping.csv', encoding='utf_8_sig')
```
