```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('F:/TensorFlow/xinjiang/traindata20230307.csv',header=0,encoding='utf_8')
df.info()
df.className.unique()
```

```python
df = df.loc[:,['className','PR2018','PR2019','PR2020','PR2021']]
df
```

```python
result = df.groupby('className').agg({'PR2018':'mean','PR2019':'mean','PR2020':'mean','PR2021':'mean'})
result.head()
```
