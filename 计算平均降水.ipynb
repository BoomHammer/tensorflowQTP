```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('F:/TensorFlow/traindata20220126.csv',header=0,encoding='utf_8')
df.info()
df.ClassName.unique()
```

```python
df = df.loc[:,['ClassName','PR2015','PR2016','PR2017','PR2018','PR2019']]
df
```

```python
result = df.groupby('ClassName').agg({'PR2015':'mean','PR2016':'mean','PR2017':'mean','PR2018':'mean','PR2019':'mean','ClassName':'size'})
result.head()
```

```python
result.to_csv("tibet_veg_prep.csv")
```
