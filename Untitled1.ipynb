```python
import pandas as pd

# 1. 读取两张CSV表格
# 注意：如果遇到编码报错，可以把 encoding='utf-8' 换成 'gbk' 或 'utf-8-sig'
df_mapping = pd.read_csv(r'G:/Eurasia/traindata20260320-mapping.csv', encoding='utf-8')
df_area = pd.read_csv(r'G:/Eurasia/area20260320.csv', encoding='utf-8')

# 2. 从 mapping 表中只提取需要的列（包括作为匹配依据的'num'列）
columns_to_keep = ['num', 'Alliance', 'Formation', 'Eng_Formation']
mapping_subset = df_mapping[columns_to_keep]

# 3. 将两张表进行合并 (Left Join)
# left_on='value' 指的是 area 表的匹配列
# right_on='num' 指的是 mapping 表的匹配列
# how='left' 保证 area 表原有的所有行和列都不受影响、不丢失
df_merged = pd.merge(df_area, mapping_subset, left_on='value', right_on='num', how='left')

# 4. （可选）删除多余的 'num' 列，因为它和 area 表里的 'value' 是一样的
if 'num' in df_merged.columns:
    df_merged = df_merged.drop(columns=['num'])

# 5. 将结果保存为新的CSV文件，不改动你原本的文件
df_merged.to_csv(r'G:/Eurasia/area_updated_3.csv', index=False, encoding='utf-8-sig')

print("处理完成！数据已保存到 area_updated.csv")
```
