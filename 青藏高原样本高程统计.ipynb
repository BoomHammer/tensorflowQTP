```python
import pandas as pd
```

```python
data_train=pd.read_csv('F:/pytorch/data/traindata20241213.csv',header=0,encoding='utf_8')
data_train.info()
```

```python
# 假设你的数据框叫df
result = data_train.groupby(['num', 'ClassName'], as_index=False)['dem1'].agg(['max', 'min']).reset_index()

# 重命名列并调整顺序
result = result[['num', 'ClassName', 'min', 'max']]

# 临时修改显示参数
with pd.option_context('display.max_rows', 200, 'display.max_columns', 10):
    print(result)  # 仅在此上下文内生效
```

```python
result.to_csv("qtpveg_dsm.csv")
```
